```python
import jax
import jax.numpy as jnp
import seaborn as sns
import matplotlib.pyplot as plt
import matplotlib.animation as animation

import treescope
treescope.basic_interactive_setup(autovisualize_arrays=True)
```

# data

```python
# Carrega o dataset direto da internet
df = sns.load_dataset('penguins')
df = df.dropna()  # Remove linhas com dados faltando

df.head()
```

```python
# Vamos usar:
# X (Feature): Tamanho da asa (flipper_length_mm)
# Y (Target):  Massa corporal (body_mass_g)
X_raw = df['flipper_length_mm'].values
Y_raw = df['body_mass_g'].values

X_mean, X_std = X_raw.mean(), X_raw.std()
Y_mean, Y_std = Y_raw.mean(), Y_raw.std()

X = (X_raw - X_mean) / X_std
Y = (Y_raw - Y_mean) / Y_std

# Converte para arrays do JAX
X = jnp.array(X)
Y = jnp.array(Y)

X, Y
```

# Modeling

```python
# Inicialização aleatória
key = jax.random.PRNGKey(42)

def initial_params(key):
    key1, key2 = jax.random.split(key)
    return {
        'w': jax.random.normal(key1),
        'b': jax.random.normal(key2)
    }

initial_params(key)
```

```python
@jax.jit
def predict(params, x):
    return params['w'] * x + params['b']

# Função de Custo: Recebe o dicionário de parâmetros
@jax.jit
def mse_loss(params, x, y):
    preds = predict(params, x)
    return jnp.mean((preds - y) ** 2)
```

```python
grad_fn = jax.grad(mse_loss, argnums=0)
```

```python
import sympy
s_w, s_b, s_x, s_y = sympy.symbols('w b x y')
# Definimos a mesma lógica do modelo, mas com símbolos
s_pred = s_w * s_x + s_b
s_loss = (s_pred - s_y)**2
# Pedimos para o SymPy derivar para nós
s_grad_w = sympy.diff(s_loss, s_w)
s_grad_b = sympy.diff(s_loss, s_b)

print("-" * 30)
print(f"Derivada (Gradiente) de W: {s_grad_w}")
print(f"Derivada (Gradiente) de b: {s_grad_b}")
print("-" * 30)
print("Isso prova que o JAX está calculando a inclinação da curva automaticamente!")
```

# Train

```python
learning_rate = 0.1
epochs = 100

print("\n🚀 Começando o treino...")
params = initial_params(key)
loss_history = []
params_history = [] # 🎞️ Vamos guardar a evolução dos pesos aqui

for i in range(epochs):
    # 1. Calcula erro
    current_loss = mse_loss(params, X, Y)
    loss_history.append(current_loss)
    
    # 2. Guarda o estado atual para o GIF (apenas a cada 2 épocas para não ficar gigante)
    if i % 2 == 0:
        params_history.append(params.copy())

    # 3. Gradiente e Atualização
    grads = grad_fn(params, X, Y)
    params = jax.tree_util.tree_map(
        lambda p, g: p - learning_rate * g, 
        params, 
        grads
    )
    
    if i % 10 == 0:
        print(f"Época {i}: Loss = {current_loss:.4f}")

print(f"🏁 Treino finalizado!")
```

# Viz

```python
def create_training_gif(history, X_norm, Y_norm, filename="mula_training.gif"):
    print("🎨 Gerando animação... (isso pode demorar uns segundos)")
    
    # Configura o plot
    fig, ax = plt.subplots(figsize=(10, 6))
    
    # Dados reais (estáticos) - Desnormalizados para o plot
    X_plot = X_raw
    Y_plot = Y_raw
    
    # Scatter plot de fundo
    ax.scatter(X_plot, Y_plot, alpha=0.5, color='blue', label='Dados Reais')
    
    # Linha inicial (vazia, vamos atualizar ela)
    line, = ax.plot([], [], color='red', linewidth=3, label='Mula Aprendendo')
    
    # Texto de status
    epoch_text = ax.text(0.02, 0.95, '', transform=ax.transAxes)
    
    # Configurações do gráfico
    ax.set_xlim(X_plot.min() - 5, X_plot.max() + 5)
    ax.set_ylim(Y_plot.min() - 500, Y_plot.max() + 500)
    ax.set_xlabel("Tamanho da Asa (mm)")
    ax.set_ylabel("Peso (g)")
    ax.set_title("Mula: Visualizando o Gradient Descent")
    ax.legend()
    ax.grid(True, alpha=0.3)

    # Função de atualização para cada frame
    def update(frame_idx):
        # Pega os pesos daquela época
        current_params = history[frame_idx]
        w_curr = current_params['w']
        b_curr = current_params['b']
        
        # Gera a linha de predição
        x_range_norm = jnp.linspace(X_norm.min(), X_norm.max(), 100)
        y_pred_norm = w_curr * x_range_norm + b_curr
        
        # Desnormaliza para plotar
        x_range_real = x_range_norm * X_std + X_mean
        y_pred_real = y_pred_norm * Y_std + Y_mean
        
        # Atualiza a linha e o texto
        line.set_data(x_range_real, y_pred_real)
        epoch_text.set_text(f'Step: {frame_idx*2}') # *2 porque salvamos a cada 2 épocas
        return line, epoch_text

    # Cria a animação
    anim = animation.FuncAnimation(
        fig, 
        update, 
        frames=len(history), 
        interval=100, # 50ms entre frames (rápido)
        blit=True
    )
    
    # Salva como GIF
    anim.save(filename, writer='pillow')
    print(f"✨ GIF salvo como: {filename}")
    plt.close() # Fecha para não duplicar output

# Chama a função mágica
create_training_gif(params_history, X, Y)
```
